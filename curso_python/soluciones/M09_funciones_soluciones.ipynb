## Ejercicio 1 · Metros cuadrados a hectáreas

Complete la función `a_hectareas(m2)` para que **devuelva** el área en hectáreas.

```python
def a_hectareas(m2):
    return m2 / 10000

print(a_hectareas(25000))
```

```python
# Comprobación
assert a_hectareas(25000) == 2.5, "25 000 m2 = 2.5 ha"
assert a_hectareas(0) == 0, "0 m2 = 0 ha"
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 2 · Área basal

Complete `area_basal(dap_cm)` para que devuelva el área basal en m²: AB = π/4 × (DAP en m)².

```python
PI = 3.141592653589793

def area_basal(dap_cm):
    dap_m = dap_cm / 100
    return PI / 4 * dap_m ** 2

print(area_basal(30))
```

```python
# Comprobación
assert abs(area_basal(30) - 0.0706858) < 1e-6, "Para DAP 30 cm, AB ≈ 0.0707 m2."
assert abs(area_basal(100) - 0.7853982) < 1e-6, "Para DAP 100 cm, AB ≈ 0.7854 m2."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 3 · Clasificador reutilizable

Escriba `clasificar_pendiente(p)` que devuelva `Suave` (< 15), `Moderada` (< 30), `Fuerte` (< 50) o `Muy fuerte`.

```python
def clasificar_pendiente(p):
    if p < 15:
        return "Suave"
    elif p < 30:
        return "Moderada"
    elif p < 50:
        return "Fuerte"
    else:
        return "Muy fuerte"

for p in [5, 15, 29.9, 50]:
    print(p, clasificar_pendiente(p))
```

```python
# Comprobación
casos = {5: "Suave", 15: "Moderada", 29.9: "Moderada", 30: "Fuerte", 50: "Muy fuerte", 80: "Muy fuerte"}
for valor, esperado in casos.items():
    obtenido = clasificar_pendiente(valor)
    assert obtenido == esperado, f"Para {valor} se esperaba '{esperado}' y se obtuvo '{obtenido}'."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 4 · Volumen con valor por defecto

Escriba `volumen(dap_cm, altura_m, ff=0.5)` que devuelva AB × altura × ff, reutilizando `area_basal`.

```python
PI = 3.141592653589793

def area_basal(dap_cm):
    return PI / 4 * (dap_cm / 100) ** 2

def volumen(dap_cm, altura_m, ff=0.5):
    return area_basal(dap_cm) * altura_m * ff

print(round(volumen(30, 12), 4))
```

```python
# Comprobación
assert abs(volumen(30, 12) - 0.4241150) < 1e-6, "Con ff por defecto (0.5) se esperaba ≈ 0.4241."
assert abs(volumen(30, 12, ff=0.7) - 0.5937610) < 1e-6, "Con ff = 0.7 se esperaba ≈ 0.5938."
print("✔ ¡Correcto! Ejercicio resuelto.")
```
