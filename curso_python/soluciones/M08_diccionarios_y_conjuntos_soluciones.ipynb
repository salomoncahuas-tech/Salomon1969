## Ejercicio 1 · Completar la ficha

Agregue la clave `pendiente` con valor 25 y cambie `area_ha` a 3.0.

```python
parcela = {"nombre": "P-07", "area_ha": 2.5}
parcela["pendiente"] = 25
parcela["area_ha"] = 3.0
print(parcela)
```

```python
# Comprobación
assert parcela == {"nombre": "P-07", "area_ha": 3.0, "pendiente": 25}, f"Se obtuvo {parcela}."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 2 · Área total de la tabla

Sume en `total` el área de todas las parcelas.

```python
parcelas = [
    {"id": "P-01", "area_ha": 2.5},
    {"id": "P-02", "area_ha": 1.8},
    {"id": "P-03", "area_ha": 3.2},
]
total = 0
for p in parcelas:
    total += p["area_ha"]
print(total)
```

```python
# Comprobación
assert abs(total - 7.5) < 1e-9, f"El total debe ser 7.5; se obtuvo {total}."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 3 · Conteo por especie

Cree el diccionario `conteo` con el número de registros de cada especie.

```python
registros = ["Faique", "Algarrobo", "Faique", "Sapote", "Faique"]
conteo = {}
for e in registros:
    conteo[e] = conteo.get(e, 0) + 1
print(conteo)
```

```python
# Comprobación
assert conteo == {"Faique": 3, "Algarrobo": 1, "Sapote": 1}, f"Se obtuvo {conteo}."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 4 · Especies compartidas

Guarde en `comunes` el conjunto de especies presentes en ambas parcelas.

```python
p1 = ["Algarrobo", "Faique", "Hualtaco", "Faique"]
p2 = ["Sapote", "Faique", "Algarrobo"]
comunes = set(p1) & set(p2)
print(comunes)
```

```python
# Comprobación
assert comunes == {"Algarrobo", "Faique"}, f"Se obtuvo {comunes}."
print("✔ ¡Correcto! Ejercicio resuelto.")
```
