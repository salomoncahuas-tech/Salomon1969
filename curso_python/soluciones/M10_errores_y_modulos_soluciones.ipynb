## Ejercicio 1 · Conversión segura

Complete `a_numero(texto)`: devuelve el número decimal o `None` si el texto no es numérico.

```python
def a_numero(texto):
    try:
        return float(texto)
    except ValueError:
        return None

print(a_numero("12.5"), a_numero("abc"))
```

```python
# Comprobación
assert a_numero("12.5") == 12.5, "'12.5' debe convertirse en 12.5"
assert a_numero("abc") is None, "'abc' debe devolver None"
assert a_numero("7") == 7.0, "'7' debe convertirse en 7.0"
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 2 · División segura

Complete `densidad(arboles, area_ha)`: devuelve árboles por hectárea, o `None` si el área es 0.

```python
def densidad(arboles, area_ha):
    try:
        return arboles / area_ha
    except ZeroDivisionError:
        return None

print(densidad(450, 0.5), densidad(10, 0))
```

```python
# Comprobación
assert densidad(450, 0.5) == 900, "450 árboles en 0.5 ha = 900 árb/ha"
assert densidad(10, 0) is None, "Con área 0 debe devolver None"
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 3 · DAP a partir de la circunferencia

En campo se mide la circunferencia a la altura del pecho (CAP). Complete `dap_desde_cap(cap_cm)` = CAP / π usando `math.pi`.

```python
import math

def dap_desde_cap(cap_cm):
    return cap_cm / math.pi

print(round(dap_desde_cap(100), 2))
```

```python
# Comprobación
import math
assert abs(dap_desde_cap(100) - 31.8310) < 1e-3, "CAP 100 cm → DAP ≈ 31.83 cm"
assert abs(dap_desde_cap(math.pi * 20) - 20) < 1e-9, "CAP de un DAP de 20 cm debe dar 20"
print("✔ ¡Correcto! Ejercicio resuelto.")
```
