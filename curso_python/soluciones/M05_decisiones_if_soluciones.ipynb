## Ejercicio 1 · Par o impar

Guarde en `tipo` el texto `par` o `impar` según el valor de `numero`.

```python
numero = 17
if numero % 2 == 0:
    tipo = "par"
else:
    tipo = "impar"
print(tipo)
```

```python
# Comprobación
assert tipo == "impar", "17 es impar."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 2 · Aprobado o desaprobado

En escala vigesimal se aprueba con 11 o más. Guarde en `estado` el texto `Aprobado` o `Desaprobado` para `nota = 10`.

```python
nota = 10
estado = "Aprobado" if nota >= 11 else "Desaprobado"
print(estado)
```

```python
# Comprobación
assert estado == "Desaprobado", "Con 10 el estado debe ser 'Desaprobado'."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 3 · Clasificar la lluvia del día

Clasifique `lluvia` en `clase` con estos umbrales de ejemplo: 0 → `Sin lluvia`; hasta 10 → `Ligera`; hasta 30 → `Moderada`; mayor a 30 → `Intensa`.

```python
lluvia = 18.5
if lluvia == 0:
    clase = "Sin lluvia"
elif lluvia <= 10:
    clase = "Ligera"
elif lluvia <= 30:
    clase = "Moderada"
else:
    clase = "Intensa"
print(clase)
```

```python
# Comprobación
assert clase == "Moderada", f"Con 18.5 mm se esperaba 'Moderada' y se obtuvo '{clase}'."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 4 · Prioridad de intervención

Asigne `prioridad`: `Alta` si pendiente > 30 y cobertura < 30; `Media` si solo se cumple una de las dos; `Baja` en otro caso.

```python
pendiente = 45
cobertura = 55
if pendiente > 30 and cobertura < 30:
    prioridad = "Alta"
elif pendiente > 30 or cobertura < 30:
    prioridad = "Media"
else:
    prioridad = "Baja"
print(prioridad)
```

```python
# Comprobación
assert prioridad == "Media", f"Solo se cumple la pendiente: se esperaba 'Media' y se obtuvo '{prioridad}'."
print("✔ ¡Correcto! Ejercicio resuelto.")
```
