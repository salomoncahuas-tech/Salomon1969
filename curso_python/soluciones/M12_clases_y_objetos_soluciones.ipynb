## Ejercicio 1 · Clase Parcela

Complete la clase `Parcela` con los atributos `nombre` y `area_ha`, y el método `area_m2()` que devuelva el área en m².

```python
class Parcela:
    def __init__(self, nombre, area_ha):
        self.nombre = nombre
        self.area_ha = area_ha

    def area_m2(self):
        return self.area_ha * 10000

p = Parcela("P-01", 2.5)
print(p.nombre, p.area_m2())
```

```python
# Comprobación
p = Parcela("P-09", 1.2)
assert p.nombre == "P-09", "El atributo nombre no se guardó."
assert abs(p.area_m2() - 12000) < 1e-9, "1.2 ha = 12 000 m2"
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 2 · Árbol con métodos

Complete `area_basal()` (π/4 × DAP² en m) y `volumen(ff=0.5)` (AB × altura × ff).

```python
import math

class Arbol:
    def __init__(self, dap_cm, altura_m):
        self.dap_cm = dap_cm
        self.altura_m = altura_m

    def area_basal(self):
        return math.pi / 4 * (self.dap_cm / 100) ** 2

    def volumen(self, ff=0.5):
        return self.area_basal() * self.altura_m * ff

a = Arbol(30, 12)
print(a.area_basal(), a.volumen())
```

```python
# Comprobación
a = Arbol(30, 12)
assert abs(a.area_basal() - 0.0706858) < 1e-6, "AB de DAP 30 cm ≈ 0.0707 m2"
assert abs(a.volumen() - 0.4241150) < 1e-6, "Volumen con ff 0.5 ≈ 0.4241 m3"
assert abs(a.volumen(ff=0.7) - 0.5937610) < 1e-6, "Volumen con ff 0.7 ≈ 0.5938 m3"
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 3 · Inventario

Complete la clase `Inventario`: `agregar(dap_cm)` guarda un DAP en la lista y `promedio()` devuelve el DAP promedio (0 si está vacío).

```python
class Inventario:
    def __init__(self):
        self.daps = []

    def agregar(self, dap_cm):
        self.daps.append(dap_cm)

    def promedio(self):
        if len(self.daps) == 0:
            return 0
        return sum(self.daps) / len(self.daps)

inv = Inventario()
for d in [20, 30, 40]:
    inv.agregar(d)
print(inv.promedio())
```

```python
# Comprobación
vacio = Inventario()
assert vacio.promedio() == 0, "Un inventario vacío debe devolver 0."
inv = Inventario()
for d in [20, 30, 40]:
    inv.agregar(d)
assert inv.daps == [20, 30, 40], "agregar() no guarda los valores."
assert inv.promedio() == 30, "El promedio de 20, 30 y 40 es 30."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 4 · Mostrar con __str__

Agregue `__str__` a la clase para que `str(Especie("Algarrobo", "Prosopis pallida"))` devuelva `Algarrobo (Prosopis pallida)`.

```python
class Especie:
    def __init__(self, comun, cientifico):
        self.comun = comun
        self.cientifico = cientifico

    def __str__(self):
        return f"{self.comun} ({self.cientifico})"

print(Especie("Algarrobo", "Prosopis pallida"))
```

```python
# Comprobación
assert str(Especie("Algarrobo", "Prosopis pallida")) == "Algarrobo (Prosopis pallida)", "Revise el formato del texto."
print("✔ ¡Correcto! Ejercicio resuelto.")
```
