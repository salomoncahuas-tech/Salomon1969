## Ejercicio 1 · Actualizar la lista de especies

Agregue `Faique` al final de la lista y elimine `Eucalipto`.

```python
especies = ["Algarrobo", "Eucalipto", "Hualtaco"]
especies.append("Faique")
especies.remove("Eucalipto")
print(especies)
```

```python
# Comprobación
assert especies == ["Algarrobo", "Hualtaco", "Faique"], f"Se obtuvo {especies}."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 2 · Resumen de alturas

Calcule `maximo`, `minimo` y `promedio` de la lista de alturas.

```python
alturas = [12.5, 9.8, 15.1, 11.0, 13.6]
maximo = max(alturas)
minimo = min(alturas)
promedio = sum(alturas) / len(alturas)
print(maximo, minimo, round(promedio, 2))
```

```python
# Comprobación
assert maximo == 15.1 and minimo == 9.8, "Revise máximo y mínimo."
assert abs(promedio - 12.4) < 1e-9, "El promedio debe ser 12.4."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 3 · De centímetros a metros

Con una comprensión de lista, cree `dap_m` con los diámetros convertidos a metros.

```python
dap_cm = [20, 35, 50]
dap_m = [d / 100 for d in dap_cm]
print(dap_m)
```

```python
# Comprobación
assert dap_m == [0.2, 0.35, 0.5], f"Se obtuvo {dap_m}."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 4 · Distancia entre dos puntos

Desempaquete las tuplas y calcule en `distancia` la distancia en metros entre los puntos con el teorema de Pitágoras: √((x2−x1)² + (y2−y1)²).

```python
p1 = (500000, 9400000)
p2 = (500300, 9400400)
x1, y1 = p1
x2, y2 = p2
distancia = ((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5
print(distancia)
```

```python
# Comprobación
assert abs(distancia - 500) < 1e-9, f"La distancia debe ser 500 m; se obtuvo {distancia}."
print("✔ ¡Correcto! Ejercicio resuelto.")
```
