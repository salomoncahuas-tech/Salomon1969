## Ejercicio 1 · Contar líneas

Guarde en `n_lineas` cuántas líneas con contenido tiene el texto.

```python
texto = """Parcela 1 instalada
Parcela 2 instalada
Parcela 3 pendiente
"""
n_lineas = len(texto.strip().split("\n"))
print(n_lineas)
```

```python
# Comprobación
assert n_lineas == 3, f"Son 3 líneas; se obtuvo {n_lineas}."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 2 · Total de árboles desde un CSV

Lea el CSV con `csv.DictReader` y sume en `total` la columna `arboles`.

```python
import csv, io

datos = """parcela,arboles
P-01,45
P-02,38
P-03,52
"""
total = 0
for fila in csv.DictReader(io.StringIO(datos)):
    total += int(fila["arboles"])
print(total)
```

```python
# Comprobación
assert total == 135, f"El total debe ser 135; se obtuvo {total}."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 3 · Promedio por especie

Calcule en `promedios` un diccionario con el DAP promedio de cada especie (redondeado a 2 decimales).

```python
import csv, io

datos = """especie,dap_cm
Algarrobo,30
Faique,18
Algarrobo,40
Faique,22
Algarrobo,35
"""
suma = {}
cuenta = {}
for fila in csv.DictReader(io.StringIO(datos)):
    e = fila["especie"]
    suma[e] = suma.get(e, 0) + float(fila["dap_cm"])
    cuenta[e] = cuenta.get(e, 0) + 1
promedios = {e: round(suma[e] / cuenta[e], 2) for e in suma}
print(promedios)
```

```python
# Comprobación
assert promedios == {"Algarrobo": 35.0, "Faique": 20.0}, f"Se obtuvo {promedios}."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 4 · Leer un JSON

Convierta el texto JSON en un diccionario y guarde en `responsable` el nombre del responsable.

```python
import json

texto = '{"brigada": 2, "responsable": "J. Pérez", "parcelas": [1, 2, 3]}'
datos = json.loads(texto)
responsable = datos["responsable"]
print(responsable)
```

```python
# Comprobación
assert responsable == "J. Pérez", f"Se obtuvo {responsable!r}."
print("✔ ¡Correcto! Ejercicio resuelto.")
```
