## Ejercicio 1 · Ficha de un árbol

Cree cuatro variables: `especie` con el texto `Algarrobo`, `dap_cm` con 32.5, `altura_m` con el entero 8 y `es_nativa` con True.

```python
especie = "Algarrobo"
dap_cm = 32.5
altura_m = 8
es_nativa = True
print(especie, dap_cm, altura_m, es_nativa)
```

```python
# Comprobación
assert especie == "Algarrobo", "especie debe ser el texto 'Algarrobo'."
assert dap_cm == 32.5 and isinstance(dap_cm, float), "dap_cm debe ser el decimal 32.5."
assert altura_m == 8 and isinstance(altura_m, int), "altura_m debe ser el entero 8."
assert es_nativa is True, "es_nativa debe ser True (sin comillas)."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 2 · Texto a número

Un formulario entregó la precipitación como texto: `"125.4"`. Conviértala a decimal en `lluvia_mm` y súmele 10.6 en `lluvia_total`.

```python
dato = "125.4"
lluvia_mm = float(dato)
lluvia_total = lluvia_mm + 10.6
print(lluvia_total)
```

```python
# Comprobación
assert isinstance(lluvia_mm, float), "lluvia_mm debe ser un float."
assert abs(lluvia_total - 136.0) < 1e-9, "lluvia_total debe ser 136.0."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 3 · Contador de árboles

Empiece con `arboles = 0`. Sume 15 (día 1), sume 23 (día 2) y reste 4 (árboles mal registrados) usando `+=` y `-=`.

```python
arboles = 0
arboles += 15
arboles += 23
arboles -= 4
print(arboles)
```

```python
# Comprobación
assert arboles == 34, f"Se esperaba 34 y se obtuvo {arboles}."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 4 · Intercambio de valores

Por error se registró `este = 9425300` y `norte = 545120`. Intercambie los valores en una sola línea.

```python
este = 9425300
norte = 545120
este, norte = norte, este
print(este, norte)
```

```python
# Comprobación
assert este == 545120 and norte == 9425300, "Los valores no se intercambiaron."
print("✔ ¡Correcto! Ejercicio resuelto.")
```
