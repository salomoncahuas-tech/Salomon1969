## Ejercicio 1 · Suma del 1 al 100

Use un bucle `for` con `range()` para sumar los números del 1 al 100 en `total`.

```python
total = 0
for i in range(1, 101):
    total += i
print(total)
```

```python
# Comprobación
assert total == 5050, f"Se esperaba 5050 y se obtuvo {total}."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 2 · Contar árboles altos

Cuente en `altos` cuántas alturas de la lista son mayores a 10 m.

```python
alturas = [8.5, 12.1, 10.0, 15.3, 9.9, 11.2]
altos = 0
for h in alturas:
    if h > 10:
        altos += 1
print(altos)
```

```python
# Comprobación
assert altos == 3, f"Son 3 (12.1, 15.3 y 11.2); se obtuvo {altos}."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 3 · ¿Cuántos años?

Un árbol mide 200 cm y crece 80 cm por año. Con `while`, calcule en `anios` cuántos años tarda en alcanzar al menos 1000 cm.

```python
altura = 200
anios = 0
while altura < 1000:
    altura += 80
    anios += 1
print(anios)
```

```python
# Comprobación
assert anios == 10, f"Se esperaban 10 años y se obtuvo {anios}."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 4 · Factorial

Calcule en `producto` el factorial de 5 (1 × 2 × 3 × 4 × 5) con un bucle.

```python
producto = 1
for i in range(1, 6):
    producto *= i
print(producto)
```

```python
# Comprobación
assert producto == 120, f"5! = 120; se obtuvo {producto}."
print("✔ ¡Correcto! Ejercicio resuelto.")
```
