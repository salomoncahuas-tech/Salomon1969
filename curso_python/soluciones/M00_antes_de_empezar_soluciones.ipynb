## Ejercicio 1 · Su primer mensaje

Guarde el texto `Hola, soy aprendiz de Python` en `mensaje` (ya está escrito el inicio) y muéstrelo con `print`.

```python
mensaje = "Hola, soy aprendiz de Python"
print(mensaje)
```

```python
# Comprobación
assert mensaje == "Hola, soy aprendiz de Python", "El texto no coincide exactamente: revise mayúsculas, coma y espacios."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 3 · Python como calculadora

En un vivero se produjeron 125 plantones en enero y 375 en febrero. Calcule el total y guárdelo en `total`.

```python
total = 125 + 375
print(total)
```

```python
# Comprobación
assert total == 500, f"Se esperaba 500 y se obtuvo {total}."
print("✔ ¡Correcto! Ejercicio resuelto.")
```
