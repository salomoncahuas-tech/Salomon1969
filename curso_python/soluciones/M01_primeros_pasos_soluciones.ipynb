## Ejercicio 1 · Superficie de una parcela

Una parcela mide 20 m por 50 m. Calcule su superficie en m² (`area_m2`) y en hectáreas (`area_ha`). Recuerde: 1 ha = 10 000 m².

```python
area_m2 = 20 * 50
area_ha = area_m2 / 10000
print(area_m2, "m2")
print(area_ha, "ha")
```

```python
# Comprobación
assert area_m2 == 1000, "La superficie en m2 debe ser 1000."
assert abs(area_ha - 0.1) < 1e-9, "La superficie en ha debe ser 0.1."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 2 · Altura promedio

Se midieron tres árboles: 12.5 m, 14.0 m y 13.6 m. Calcule la altura promedio en `promedio`.

```python
promedio = (12.5 + 14.0 + 13.6) / 3
print(round(promedio, 2))
```

```python
# Comprobación
assert abs(promedio - 13.3667) < 0.001, "Revise los paréntesis: primero se suma y luego se divide."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 3 · Cajas de plantones

Hay 1100 plantones y cada caja lleva 40. Calcule cuántas cajas se llenan completas (`cajas_llenas`) y cuántos plantones sobran (`sobrantes`).

```python
cajas_llenas = 1100 // 40
sobrantes = 1100 % 40
print(cajas_llenas, sobrantes)
```

```python
# Comprobación
assert cajas_llenas == 27, "Las cajas llenas deben ser 27."
assert sobrantes == 20, "Deben sobrar 20 plantones."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 4 · De Celsius a Fahrenheit

Convierta 15 °C a grados Fahrenheit con la fórmula F = C × 9/5 + 32. Guarde el resultado en `fahrenheit`.

```python
celsius = 15
fahrenheit = celsius * 9 / 5 + 32
print(fahrenheit)
```

```python
# Comprobación
assert abs(fahrenheit - 59) < 1e-9, "15 °C equivalen a 59 °F."
print("✔ ¡Correcto! Ejercicio resuelto.")
```
