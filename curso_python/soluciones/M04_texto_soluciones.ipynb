## Ejercicio 1 · Nombre completo

Una `nombre` y `apellido` con un espacio en medio y guarde el resultado en `completo`.

```python
nombre = "Ana"
apellido = "Torres"
completo = f"{nombre} {apellido}"
print(completo)
```

```python
# Comprobación
assert completo == "Ana Torres", f"Se esperaba 'Ana Torres' y se obtuvo '{completo}'."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 2 · Limpiar un dato de campo

El texto `"   ALGARROBO "` viene con espacios y en mayúsculas. Guarde en `limpio` la versión `Algarrobo`.

```python
dato = "   ALGARROBO "
limpio = dato.strip().capitalize()
print(f"[{limpio}]")
```

```python
# Comprobación
assert limpio == "Algarrobo", f"Se obtuvo '{limpio}'."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 3 · Descomponer un código

Del código `"PI-ALG-0045"`, guarde en `especie` el texto `ALG` y en `numero` el entero `45`.

```python
codigo = "PI-ALG-0045"
partes = codigo.split("-")
especie = partes[1]
numero = int(partes[2])
print(especie, numero)
```

```python
# Comprobación
assert especie == "ALG", "especie debe ser 'ALG'."
assert numero == 45 and isinstance(numero, int), "numero debe ser el entero 45."
print("✔ ¡Correcto! Ejercicio resuelto.")
```

## Ejercicio 4 · Línea de reporte

Construya en `reporte` el texto exacto `Parcela 3: 12.50 ha` usando una f-string con 2 decimales.

```python
parcela = 3
area = 12.5
reporte = f"Parcela {parcela}: {area:.2f} ha"
print(reporte)
```

```python
# Comprobación
assert reporte == "Parcela 3: 12.50 ha", f"Se obtuvo '{reporte}'."
print("✔ ¡Correcto! Ejercicio resuelto.")
```
