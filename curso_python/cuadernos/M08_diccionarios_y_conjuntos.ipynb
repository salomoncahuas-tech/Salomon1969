## 1. ¿Qué es un diccionario?

Un **diccionario** guarda pares **clave: valor** entre llaves `{ }`. En lugar de acceder por posición, se accede **por nombre** (clave). Es como una ficha de campo: cada campo tiene un nombre y un valor.

- `dic["clave"]` obtiene el valor (error si no existe).
- `dic.get("clave", defecto)` obtiene el valor o un valor por defecto si no existe.

```python
arbol = {
    "codigo": "A-001",
    "especie": "Algarrobo",
    "dap_cm": 32.5,
    "altura_m": 9.0,
}
print(arbol["especie"])
print(arbol.get("estado", "Por verificar"))
```

## 2. Modificar, agregar y eliminar

- Asignar a una clave existente **modifica** su valor; a una clave nueva, la **agrega**.
- `del dic["clave"]` elimina un par.
- `"clave" in dic` verifica si existe la clave.

```python
parcela = {"nombre": "P-01", "area_ha": 2.5}
parcela["area_ha"] = 2.8          # modifica
parcela["pendiente"] = 25         # agrega
print(parcela)
del parcela["pendiente"]
print("pendiente" in parcela)
```

## 3. Recorrer un diccionario

- `.keys()` → claves, `.values()` → valores, `.items()` → pares (clave, valor).
- Lo más usado es recorrer con `.items()`.

```python
lluvia_mensual = {"ene": 45.2, "feb": 120.8, "mar": 210.5}
for mes, mm in lluvia_mensual.items():
    print(f"{mes}: {mm:>6.1f} mm")
print("Total:", sum(lluvia_mensual.values()))
```

## 4. Una tabla: lista de diccionarios

Una lista de diccionarios equivale a una **tabla**: cada diccionario es una **fila** y cada clave una **columna**. Esta estructura es el puente hacia los archivos CSV (Módulo 11) y pandas (Módulo 13).

*Datos ficticios de ejemplo.*

```python
parcelas = [
    {"id": "P-01", "area_ha": 2.5, "especie": "Algarrobo"},
    {"id": "P-02", "area_ha": 1.8, "especie": "Faique"},
    {"id": "P-03", "area_ha": 3.2, "especie": "Algarrobo"},
]
total = 0
for p in parcelas:
    print(p["id"], p["area_ha"])
    total += p["area_ha"]
print("Área total:", round(total, 2), "ha")
```

## 5. Contar frecuencias

Un diccionario es ideal para contar cuántas veces aparece cada valor. Patrón: `conteo[x] = conteo.get(x, 0) + 1`.

```python
registros = ["Algarrobo", "Faique", "Algarrobo", "Hualtaco", "Algarrobo", "Faique"]
conteo = {}
for especie in registros:
    conteo[especie] = conteo.get(especie, 0) + 1
print(conteo)
```

## 6. Conjuntos (set)

Un **conjunto** guarda valores **sin repetir y sin orden**. Útil para obtener valores únicos y comparar grupos:

- **Unión** `a | b`: elementos en a o en b.
- **Intersección** `a & b`: elementos presentes en ambos.
- **Diferencia** `a - b`: elementos en a pero no en b.

```python
parcela_1 = {"Algarrobo", "Faique", "Hualtaco"}
parcela_2 = {"Algarrobo", "Palo santo", "Faique"}
print("En ambas:", sorted(parcela_1 & parcela_2))
print("Solo en la 1:", sorted(parcela_1 - parcela_2))
print("Todas:", sorted(parcela_1 | parcela_2))
print(set([1, 1, 2, 3, 3]))
```
