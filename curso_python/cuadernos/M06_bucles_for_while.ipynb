## 1. El bucle for y range()

Un **bucle** repite instrucciones. `for` recorre una secuencia de valores, uno por uno, y en cada vuelta la variable toma el siguiente valor.

`range()` genera secuencias de números enteros:

| Expresión | Valores |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 |

Igual que el `if`, la línea termina en `:` y el bloque lleva sangría.

```python
for i in range(1, 6):
    print("Parcela", i)
print("Recorrido terminado")
```

## 2. Recorrer una lista o un texto

Una **lista** (se estudia a fondo en el Módulo 7) es una colección de valores entre corchetes `[ ]`. `for` puede recorrerla directamente. `enumerate()` entrega además la posición.

```python
especies = ["Algarrobo", "Faique", "Hualtaco"]
for especie in especies:
    print("Especie:", especie)

for n, especie in enumerate(especies, start=1):
    print(n, especie)
```

## 3. Acumuladores y contadores

Patrón muy usado:

1. **Antes** del bucle, crear la variable en 0.
2. **Dentro** del bucle, sumarle algo en cada vuelta.
3. **Después** del bucle, usar el resultado.

```python
alturas = [8.5, 12.1, 10.0, 15.3, 9.9]

total = 0          # acumulador
mayores = 0        # contador
for h in alturas:
    total += h
    if h > 10:
        mayores += 1

print("Suma de alturas:", round(total, 2))
print("Promedio:", round(total / len(alturas), 2))
print("Árboles de más de 10 m:", mayores)
```

## 4. El bucle while

`while` (mientras) repite **mientras la condición sea verdadera**. Se usa cuando no se sabe de antemano cuántas vueltas habrá.

**Cuidado:** dentro del bucle algo debe cambiar para que la condición llegue a ser falsa; si no, el bucle sería infinito (en Jupyter se detiene con el botón ■ *Interrupt*).

```python
altura_cm = 50          # plantón recién instalado (valores de ejemplo)
crecimiento_anual = 60
anios = 0
while altura_cm < 300:
    altura_cm += crecimiento_anual
    anios += 1
    print(f"Año {anios}: {altura_cm} cm")
print(f"Supera los 3 m en {anios} años")
```

## 5. break y continue

- `break` **termina** el bucle inmediatamente.
- `continue` **salta** a la siguiente vuelta sin ejecutar el resto del bloque.

```python
lecturas = [12, 15, -1, 18, 999, 20]
for valor in lecturas:
    if valor < 0:
        print("Lectura inválida, se omite")
        continue
    if valor == 999:
        print("Código de fin de datos")
        break
    print("Lectura:", valor)
```
