## 1. ¿Qué es una función?

Ya ha usado funciones: `print()`, `len()`, `round()`. Una función es un **bloque de código con nombre** que realiza una tarea y se puede **reutilizar** cuantas veces se necesite.

```
def nombre_funcion(parametro1, parametro2):
    instrucciones
    return resultado
```

- `def` define la función (no la ejecuta).
- La función se ejecuta al **llamarla**: `nombre_funcion(valor1, valor2)`.

````python
def saludar(nombre):
    print(f"Hola, {nombre}. Bienvenido al curso.")

saludar("Ana")
saludar("Luis")
````

## 2. Devolver resultados con return

`return` **entrega** un valor a quien llamó la función, para guardarlo o seguir operando con él. `print` solo lo **muestra**.

Regla práctica: las funciones de cálculo deben usar `return`, no `print`.

````python
def area_basal(dap_cm):
    """Área basal (m²) a partir del DAP en cm: AB = π/4 × DAP²."""
    pi = 3.141592653589793
    dap_m = dap_cm / 100
    return pi / 4 * dap_m ** 2

ab = area_basal(30)
print(round(ab, 4), "m2")

total = area_basal(30) + area_basal(25) + area_basal(40)
print("AB total:", round(total, 4), "m2")
````

## 3. Parámetros por defecto y argumentos con nombre

Un parámetro puede tener un **valor por defecto**: si no se entrega, se usa ese. Al llamar, puede indicar los argumentos **por nombre**, lo que hace el código más claro.

Una función también puede devolver **varios valores** (como tupla).

````python
def volumen(dap_cm, altura_m, ff=0.5):
    """Volumen (m³) = AB × altura × factor de forma (ff).
    ff = 0.5 es un valor supuesto para este ejemplo."""
    ab = 3.141592653589793 / 4 * (dap_cm / 100) ** 2
    return ab * altura_m * ff

print(round(volumen(30, 12), 4))
print(round(volumen(30, 12, ff=0.6), 4))
print(round(volumen(altura_m=12, dap_cm=30), 4))

def min_max(datos):
    return min(datos), max(datos)

menor, mayor = min_max([4, 9, 2, 7])
print(menor, mayor)
````

## 4. Documentar: docstrings

El texto entre comillas triples justo debajo de `def` es el **docstring**: explica qué hace la función, qué recibe y qué devuelve. `help(funcion)` lo muestra.

Buenas prácticas: nombres de funciones con verbos o sustantivos claros (`calcular_volumen`), una sola tarea por función y funciones cortas.

## 5. Alcance de las variables

Las variables creadas **dentro** de una función son **locales**: solo existen mientras la función se ejecuta. Para sacar un valor de la función, use `return`.

````python
def calcular():
    resultado_local = 42
    return resultado_local

valor = calcular()
print(valor)
# print(resultado_local)  # daría NameError: no existe fuera de la función
````

## 6. Funciones como herramienta de orden: key y lambda

Algunas funciones reciben **otra función** como argumento. Por ejemplo, `sorted(lista, key=...)` ordena según lo que devuelva esa función.

`lambda` crea una función pequeña de una sola línea, sin nombre: `lambda x: x["dap"]`.

````python
arboles = [
    {"id": "A1", "dap": 32.5},
    {"id": "A2", "dap": 18.0},
    {"id": "A3", "dap": 45.2},
]
ordenados = sorted(arboles, key=lambda a: a["dap"], reverse=True)
for a in ordenados:
    print(a["id"], a["dap"])
````

# Glosario del módulo

| Término | Significado |
|---|---|
| **def** | Palabra clave para definir una función. |
| **Parámetro / argumento** | Parámetro: nombre en la definición. Argumento: valor entregado al llamar. |
| **return** | Instrucción que devuelve un valor y termina la función. |
| **Docstring** | Texto de documentación al inicio de una función. |
| **lambda** | Función anónima de una sola expresión. |
| **Alcance (scope)** | Zona del programa donde una variable existe. |
